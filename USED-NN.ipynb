```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
```

```python
df_train['brand'] = label.fit_transform(df_train['brand'])
df_test['brand'] = label.transform(df_test['brand'])
```

```python
df_train['fuel_type'] = label.fit_transform(df_train['fuel_type'])
df_test['fuel_type'] = label.transform(df_test['fuel_type'])
```

```python
df_train['engine'] = label.fit_transform(df_train['engine'])
df_test['engine'] = label.transform(df_test['engine'])
```

```python
df_train['transmission'] = label.fit_transform(df_train['transmission'])
df_test['transmission'] = label.transform(df_test['transmission'])
```

```python
df_train['ext_col'] = label.fit_transform(df_train['ext_col'])
df_test['ext_col'] = label.transform(df_test['ext_col'])
```

```python
df_train['int_col'] = label.fit_transform(df_train['int_col'])
df_test['int_col'] = label.transform(df_test['int_col'])
```

```python
df_train['accident'] = label.fit_transform(df_train['accident'])
df_test['accident'] = label.transform(df_test['accident'])
```

```python
df_train['clean_title'] = label.fit_transform(df_train['clean_title'])
df_test['clean_title'] = label.transform(df_test['clean_title'])
```

```python
df_train.corr()['price']
```

```python
X_train = df_train.drop('price', axis=1)
y_train = df_train['price']
X_test = df_test.copy()
```

```python
df_train.shape
```

```python
from keras import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
```

```python
model = Sequential()
```

```python
model.add(Dense(units=26, activation='relu',  use_bias=False , input_shape=(10, )))
model.add(Dense(units=13, activation='relu', use_bias=False))
model.add(Dense(units=1, activation='linear'))
```

```python
model.compile(optimizer=Adam(learning_rate=0.0001), loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
y_train
```

```python
model.fit(X_train , y_train, epochs= 20)
```

```python
y_pred = model.predict(X_test)
```

```python
df_sub['price'] = y_pred
```

```python
df_sub.to_csv('sub_NN.csv', index=False, index_label=False)
```
